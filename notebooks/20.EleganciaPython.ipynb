O código "Pythonic" refere-se ao uso de idioms e construções da linguagem que a tornam mais legível, concisa e eficiente, alinhando-se à filosofia do Zen do Python ("Simples é melhor que complexo", "Legibilidade conta").

## 1.Iterando Sobre Sequências

```python
quadrados = [x**2 for x in range(10) if x % 2 == 0]
```

```python
quadrados
```

Não Pythonic

```python
items = ["maçã", "banana", "laranja"]

i = 0
while i < len(items):
    print(i, items[i])
    i += 1

# Ou, de forma um pouco melhor, mas ainda não ideal:
for i in range(len(items)):
    print(i, items[i])
```

Pythonicamente, podemos usar a função `enumerate` para obter o mesmo resultado de forma mais elegante:
for index, item in enumerate(items):

```python
items = ["maçã", "banana", "laranja"]

for i, item in enumerate(items):
    print(i, item)
```

### 2.List Comprehensions (Compreensões de Lista)
- Não Pythonic

```python
quadrados = []
for i in range(10):
    quadrados.append(i * i)
```

```python
quadrados
```

### Pythonic

Uma list comprehension expressa a mesma lógica em uma única linha, de forma mais clara e eficiente.



```python
quadrados = [i * i for i in range(10)]
quadrados
```

### 3.Trocando Variáveis

####  Não Pythonic
 - O método tradicional, usando uma variável temporária, funciona, mas requer uma linha extra de código.



```python
a = 5
b = 10

temp = a
a = b
b = temp
```

```python
print("a:", a)
print("b:", b)
```

#### Pythonic
  - Python permite a troca de variáveis em uma única linha através do desempacotamento de tuplas, o que é mais conciso e expressivo.



```python
a = 5
b = 10

a, b = b, a
```

```python
print("a:", a)
print("b:", b)
```

### 4.Verificando Contêineres Vazios

### Não Pythonic

- Comparar o comprimento (len()) com 0 funciona, mas não é a forma mais idiomática.



```python
minha_lista = []
if len(minha_lista) == 0:
    print("A lista está vazia.")
```

### Pythonic
  - Em Python, contêineres vazios são avaliados como False em um contexto booleano. Esta é a maneira preferida de verificar se estão vazios.



```python
minha_lista = []
if not minha_lista:
    print("A lista está vazia.")
```

### 5. Manipulação de Arquivos.
 - Gerenciar recursos como arquivos de forma segura é crucial para evitar vazamentos de recursos.

Não Pythonic
  - Esquecer de fechar um arquivo (arquivo.close()) ou não lidar com exceções pode causar problemas.



```python
arquivo = open("dados.txt", "w")
arquivo.write(" Python é ótimo! By Luciana Sampaio ")
# O arquivo pode não ser fechado se ocorrer um erro antes de close()
arquivo.close()
```

Pythonic
- A declaração with garante que o arquivo seja fechado automaticamente, mesmo que ocorram exceções. Isso é conhecido como gerenciamento de contexto.



```python
with open("Medium.txt", "w") as arquivo:
    arquivo.write("https://medium.com/@luciana.sampaio84")
# O arquivo é fechado automaticamente aqui
```

#### 6. Mesclando Dicionários
- A partir do Python 3.9, existe uma maneira muito elegante de mesclar dicionários.

Não Pythonic (ou pré-Python 3.9)
  - Antes, era comum usar o método update() ou desempacotamento com **.



```python
d1 = {"a": 1, "b": 2}
d2 = {"b": 3, "c": 4}

d_mesclado = d1.copy()
d_mesclado.update(d2)

# Ou com desempacotamento (Python 3.5+)
d_mesclado_2 = {**d1, **d2}
```

```python
print(d_mesclado_2)
```

 Pythonic (Python 3.9+)
- O operador de união (|) foi introduzido para dicionários, oferecendo uma sintaxe limpa e clara para a mesclagem.

```python
d1 = {"a": 1, "b": 2}
d2 = {"b": 3, "c": 4}

d_mesclado = d1 | d2
```

```python
print(d_mesclado_2)
```
