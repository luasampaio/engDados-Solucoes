```python
import csv, io, codecs, pandas as pd

path = r"E:\engDados-Solucoes\datasets\fifa19.csv"
# tente cp1252/latin1 primeiro
with open(path, "rb") as f:
    raw = f.read(2048)
for enc in ["cp1252", "latin1", "utf-8-sig", "utf-16"]:
    try:
        sample = raw.decode(enc, errors="strict")
        dialect = csv.Sniffer().sniff(sample, delimiters=";,|\t,")
        sep = dialect.delimiter
        df = pd.read_csv(path, sep=sep, encoding=enc, engine="python")
        break
    except Exception:
        continue
```

```python
import pyarrow as pa

# Criando um Array
data = pa.array([1, 2, 3, 4, 5])
print("Arrow Array:", data)

# Criando uma Table
table = pa.table({
    'column1': pa.array([1, 2, 3]),
    'column2': pa.array(['a', 'b', 'c'])
})
print("Arrow Table:", table)
```

```python
import pandas as pd
import pyarrow as pa

df = pd.DataFrame({'col1': range(5), 'col2': ['a', 'b', 'c', 'd', 'e']})

# Convertendo DataFrame pandas para tabela PyArrow (colunar)
arrow_table = pa.Table.from_pandas(df)

# Após operações em outras linguagens/processos, convertendo de volta para pandas sem cópia
df2 = arrow_table.to_pandas()
```
